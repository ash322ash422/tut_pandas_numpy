```python
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('dark_background')
```

## 2 variable: Positive Correlation

```python
X = [1,  2,  3,  4,  5] # hours studied
Y = [20, 30, 50, 40, 60] # scores out of 100

plt.figure(figsize=(8,5))
plt.scatter(X, Y, marker='x', color="gold")   

plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(0, 8)
plt.ylim(0, 100)

plt.show()
```

```python
df = pd.DataFrame({'X': X,
                   'Y': Y}
)

print("\nCovariance:\n", df.cov())
print("\nCorrelation:\n", df.corr())
```

```python
from sklearn.datasets import fetch_california_housing
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
data = fetch_california_housing()

# Create DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)
cols_to_use = [
    "MedInc", "HouseAge", "AveRooms", "AveBedrms",
    "Population", "AveOccup", 
]
df = df[cols_to_use] 
df["MedHouseValue"] = data.target

print(df.head())

# Correlation matrix
corr = df.corr()

# Correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix - California Housing Dataset")
plt.show()
```
