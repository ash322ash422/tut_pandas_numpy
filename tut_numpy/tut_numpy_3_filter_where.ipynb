```python
import numpy as np
```

```python
# filter: lets filter elements from numpy array based on condition

cont = [4, 7, 2, 3, 9, 3, 6, 2, 1, 7] # water contaminant reading
a = np.array(cont)

# Filter: select elements equal to 3
mask = (a == 3)
# print(mask)
print(a[mask])

# or use this, but sometimes not easy to understand for beginners
# print(a[a == 3]) 
```

```python
# Filter: All elements not equal to 3

cont = [4, 7, 2, 3, 9, 3, 6, 2, 1, 7] # water contaminant reading
a = np.array(cont)

mask = (a != 3)
print(a[mask]) 
```

```python
# Filter: All elem <= 2 # HEALTHY water

cont = [4, 7, 2, 3, 9, 3, 6, 2, 1, 7] # water contaminant reading
a = np.array(cont)

mask = (a <= 2)
print(a[mask])     
```

```python
# Filter: All elem > 2 # UNHEALTHY water

mask = (a > 2)
print(a[mask])     
```

```python
# Filter: All elem > 2 and < 5 # MODERATE
cont = [4, 7, 2, 3, 9, 3, 6, 2, 1, 7] # water contaminant reading
a = np.array(cont)

mask = (a > 2) & (a < 5)
print(a[mask]) 
```

```python
# select elements less than 2 or greater than 4:
cont = [4, 7, 2, 3, 9, 3, 6, 2, 1, 7] # water contaminant reading
a = np.array(cont)

mask = (a < 2) | (a > 4)
print(a[mask])
```

```python
# filter and replace: Replace values less than 20 with 0
arr = np.array([5, 12, 18, 25, 30, 3, 45])
 
arr[arr < 20] = 0
print(arr) 
```

```python
# Application of filter
student_scores = np.array([85, 42, 67, 90, 55, 38, 77, 95, 60])
print("Student Scores:", student_scores)

# Pass = score >= 60
passed = student_scores[student_scores >= 60]
failed = student_scores[student_scores < 60]

print("Passed Students:", passed)
print("Failed Students:", failed)
```

### where()
- np.where(condition):  find the indices of elements in an array that satisfy certain condition. It can also be used to create new arrays based on conditions
- np.where(condition, x, y)  → Returns x if condition is True, else y.

```python
# where(): find all elements that satisfy some condition, say elements > 25

scores = np.array([78, 45, 62, 33, 90, 50, 20])

# Indices where condition is true
indices = np.where(scores >= 50)

print(indices)      # [0, 2, 4, 5]

print(scores[indices]) # [78 62 90 50]
```

```python
# where() Application1: Assigning Letter Grades Based on Scores: >= 50 then pass otherwise fail

scores = np.array([78, 45, 62, 33, 90, 50, 20]) # Student scores

results = np.where(scores >= 50, "Pass", "Fail") # Use np.where to assign "Pass" or "Fail"

print("Scores:", scores)
print("Results:", results)
```

```python
# Application2: Daily temperatures (°C) recorded over a week
temperatures = np.array([28, 33, 31, 25, 29, 35, 30])

temp_classification = np.where(temperatures > 30, 'Greater than 30', '30 or less')

print("Temperature readings:", temperatures)
print("Classification:", temp_classification)
```

```python
print("^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^")

# Application 3: # Temperatures (°C) recorded hourly in 3 rooms over 4 hours
# You want another numpy array tat displays Alert or Normal based on temp

# Create a 2D NumPy array
temps = np.array([
    # Room1, Room2, Room3
    [22,      26,     24],  # Hour 1
    [23,      27,     26],  # Hour 2
    [25,      28,     23],  # Hour 3
    [21,      24,     27]   # Hour 4
])

# Flag temperatures above 25°C as "Alert", else "Normal"
status = np.where(temps > 25, "Alert", "Normal")

print("Temperature readings:\n", temps)
print("\nStatus:\n", status)
```

```python
# Filtering with np.isin()
# Used to filter elements that exist in another list or array.

grades = np.array(["A", "B", "C", "D", "B", "F", "B", "D", "A"])
select_grades = ["A", "B"]

mask = np.isin(grades, select_grades)
selected = grades[mask]

print("Grades:", grades)
print("Filtered (A or B):", selected)
```
