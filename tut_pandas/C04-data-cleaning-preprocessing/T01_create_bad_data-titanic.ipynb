```python
import pandas as pd
import numpy as np

df = pd.read_csv("data_titanic_kaggle.csv")
df.head()
```

```python
### 2.1 Introduce missing values

df.loc[5:15, "age"] = np.nan
df.loc[20:25, "fare"] = np.nan
```

```python
### 2.2 Introduce duplicates

df = pd.concat([df, df.iloc[:10]], ignore_index=True)
```

```python
### 2.3 Introduce wrong data types

df["pclass"] = df["pclass"].astype(str)
```

```python
### 2.4 Introduce outliers

df.loc[30, "fare"] = 1000
df.loc[40, "age"] = 200
```

```python

# Randomly select some rows
male_idx = df[df['sex'] == 'male'].sample(n=10, random_state=42).index
female_idx = df[df['sex'] == 'female'].sample(n=10, random_state=42).index

# Introduce inconsistencies
df.loc[male_idx, 'sex'] = 'MALE'
df.loc[female_idx, 'sex'] = 'Female'
```

```python
df.to_csv("bad_data_titanic.csv", index=False)
```
