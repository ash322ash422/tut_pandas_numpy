# First Look at the Dataset

**`tips` dataset** (restaurant bills & tips).

* head(), tail()
* shape
* info()


```python
## import library

import pandas as pd
```

```python
df = pd.read_csv("data_tips.csv")
```

### 1. View first few rows

```python
print(df.head())
```

* Shows column names
* Helps understand what each row represents
* Default = first **5 rows**


```python
# first 10 rows

print(df.head(10))
```

### 2. View last few rows

```python
print(df.tail())
```

**Why this matters**

* Catch weird values at the end
* Useful when data is appended manually


### 3. Dataset shape (rows, columns)

```python
print(df.shape)
```

**Interpretation**

* 244 records or instances(customers)
* 7 features (columns)


### 4. Dataset overview


```python
print(df.info())
```

**Provides info on**
    
* Column names
* Data types: float64 means decimal values, object means text values, int64 means whole number 
* Missing values
* Memory usage
