# Descriptive Statistics & Summary

```python
## import library

import pandas as pd
```

```python
df = pd.read_csv("data_tips.csv")
```

### 1. Basic statistics

```python
print(df.describe())
```

Covers:

* Mean
* Std
* Min / Max
* Quartiles


### Mean vs Median

```python
print(df["total_bill"].mean())
print(df["total_bill"].median())
```

#### Key point:

* Mean is sensitive to outliers
* Median is robust


### Standard Deviation (Spread)

```python
print(df["tip"].std())
print(df["total_bill"].std())
```

#### Key point

> “Higher std = more variability”


### 2. Distribution intuition (Quick Visual)

```python
import matplotlib.pyplot as plt

plt.hist(df["total_bill"], bins=20)
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

#### Questions and Observations:

* Is it skewed?
* Are there outliers?
