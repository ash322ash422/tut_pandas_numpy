# Descriptive Statistics & Summary

```python
## import library

import pandas as pd
```

```python
df = pd.read_csv("data_tips.csv")
```

```python
df
```

### 1. Basic statistics

```python
print(df.describe())
```

Covers:

* Mean
* Std
* Min / Max
* Quartiles


### Mean vs Median

```python
print(df["total_bill"].mean())
print(df["total_bill"].median())
```

#### Key point:

* Mean is sensitive to outliers
* Median is robust


### Standard Deviation (Spread)

```python
print(df["tip"].std())
print(df["total_bill"].std())
```

#### Key point

> “Higher std = more variability”


### 2. Distribution intuition (Quick Visual)

```python
df["total_bill"].hist(bins=20)

# Add labels and title using pandas plotting interface
ax = df["total_bill"].hist(bins=20)
ax.set_title("Distribution of Total Bill")
ax.set_xlabel("Total Bill")
ax.set_ylabel("Frequency")
```

#### Questions and Observations:

* Is it skewed?
* Are there outliers?

```python
### Boxplot intuition
```

```python
ax = df.boxplot(column="total_bill")

ax.set_title("Boxplot of Total Bill")
# ax.set_xlabel("Total Bill")
```

#### Key points
