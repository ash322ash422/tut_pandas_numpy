# Understanding Columns & Data Types 

```python
## import library

import pandas as pd
```

```python
df = pd.read_csv("data_tips.csv")
```

```python
df
```

### 1. Identify numeric vs categorical columns

```python
print(df.dtypes)
```

### Note:
By defualt:

- Numerics data types are float64 (for decimal values) or int64 (for whole numbers)
- Text data is object


### Numeric Columns

```python
# col = ['total_bill', 'tip', 'size']
col = df.select_dtypes(include="number").columns

print(col)
```

### Categorical Columns

```python
col = df.select_dtypes(include="object").columns # object or category

print(col)
```

### Convert object → category (Best Practice)

```python
# To save  memory
df["day"] = df["day"].astype("category")
df["time"] = df["time"].astype("category")
```

```python
# Check again:

print(df.dtypes)
```

### ML Impact of Data Types

| Data Type  | ML Impact        |
| ---------- | ---------------- |
| Numeric    | Used directly    |
| Object     | Must encode      |
| Category   | Memory efficient |
| Wrong type | Model errors     |


> “ML models don’t understand text — they understand numbers.”
