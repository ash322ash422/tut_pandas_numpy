````python
Great idea—weather data is **one of the best real-world time series examples** 🌦️

I’ll give you:
✅ A **synthetic but realistic weather dataset**
✅ Full **pandas time series analysis**
✅ Structured for teaching (notebook style)

---

# 📘 Weather Time Series Analysis with Pandas

## 1️⃣ Create Weather Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create date range (6 months of daily data)
dates = pd.date_range(start='2024-01-01', periods=180, freq='D')

# Create realistic weather data
np.random.seed(42)

temperature = 20 + 10*np.sin(np.linspace(0, 3*np.pi, 180)) + np.random.normal(0, 2, 180)
humidity = 60 + 20*np.random.rand(180)
rainfall = np.random.choice([0, 0, 0, 5, 10, 20], size=180)

# Create DataFrame
df = pd.DataFrame({
    'date': dates,
    'temperature': temperature,
    'humidity': humidity,
    'rainfall': rainfall
})

df.head()
```

👉 Teach:

* Temperature has **seasonality**
* Rainfall is **random events**

---

## 2️⃣ Set Date Index

```python
df.set_index('date', inplace=True)
```

---

## 3️⃣ Basic Visualization

```python
df[['temperature', 'humidity']].plot(figsize=(10,5), title="Weather Trends")
plt.show()
```

👉 Ask:

* Is temperature cyclic?
* Is humidity stable?

---

## 4️⃣ Resampling (Weekly / Monthly)

```python
# Weekly average temperature
weekly_temp = df['temperature'].resample('W').mean()

weekly_temp.plot(title="Weekly Avg Temperature")
plt.show()
```

```python
# Monthly rainfall total
monthly_rain = df['rainfall'].resample('M').sum()

monthly_rain.plot(title="Monthly Rainfall")
plt.show()
```

👉 Teach:

* Avg for temperature
* Sum for rainfall

---

## 5️⃣ Rolling Window (Smooth Weather)

```python
df['temp_rolling_7'] = df['temperature'].rolling(window=7).mean()

df[['temperature', 'temp_rolling_7']].plot(title="7-Day Smoothed Temperature")
plt.show()
```

👉 Real-world:

* Used in weather reports

---

## 6️⃣ Time-Based Filtering

```python
# Get February data
feb_data = df.loc['2024-02']

feb_data
```

---

## 7️⃣ Feature Extraction (Important for ML)

```python
df['month'] = df.index.month
df['day'] = df.index.day
df['day_name'] = df.index.day_name()

df.head()
```

👉 Use cases:

* Predict temperature
* Rain prediction

---

## 8️⃣ Lag Features (Weather Dependency)

```python
df['temp_lag_1'] = df['temperature'].shift(1)
df['temp_lag_3'] = df['temperature'].shift(3)

df.head()
```

👉 Teach:

* Yesterday’s weather affects today

---

## 9️⃣ Detect Rainy Days

```python
df['is_rainy'] = df['rainfall'] > 0

df[['rainfall', 'is_rainy']].head(15)
```

👉 Real-world:

* Binary classification feature

---

## 🔟 Rolling Rainfall (Rain Trend)

```python
df['rain_rolling_7'] = df['rainfall'].rolling(7).sum()

df[['rainfall', 'rain_rolling_7']].plot(title="7-Day Rainfall Accumulation")
plt.show()
```

👉 Teach:

* Flood risk / accumulation

---

## 1️⃣1️⃣ Correlation Between Features

```python
df[['temperature', 'humidity', 'rainfall']].corr()
```

👉 Ask:

* Does rain affect temperature?

---

## 1️⃣2️⃣ Differencing (Remove Trend)

```python
df['temp_diff'] = df['temperature'].diff()

df[['temperature', 'temp_diff']].plot(title="Temperature Differencing")
plt.show()
```

---

# 🎯 Key Teaching Concepts Covered

| Concept             | Example           |
| ------------------- | ----------------- |
| Trend               | Temperature curve |
| Seasonality         | Sine wave pattern |
| Noise               | Random variation  |
| Resampling          | Weekly/monthly    |
| Rolling             | 7-day average     |
| Lag                 | Yesterday’s temp  |
| Feature engineering | Month, rain flag  |

---

# 🧑‍🏫 Teaching Flow (Very Effective)

1. Show raw data → ask patterns
2. Plot → identify seasonality
3. Resample → weekly/monthly
4. Rolling → smoothing
5. Lag → dependency
6. Features → ML connection

---

# 💡 Classroom Insight

Ask students:

👉 “Can we predict tomorrow’s temperature using past 3 days?”

This naturally leads to:
➡️ Time series forecasting
➡️ Machine learning

---

If you want next:
✅ I can turn this into a **mini project: Weather Prediction Model**
✅ Or give **student exercises + assignment questions**
✅ Or connect this to **real API data (live weather)**
````
