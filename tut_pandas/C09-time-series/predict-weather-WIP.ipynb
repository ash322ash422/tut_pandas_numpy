````python
**Complete prediction pipeline** 

We’ll predict:
👉 **Tomorrow’s temperature**

---

# 📘 Weather Prediction Model (End-to-End)

## 1️⃣ Problem Statement

> Use past weather data to predict **next day temperature**

---

## 2️⃣ Prepare Dataset (Reuse Previous Data)

```python
import pandas as pd
import numpy as np

# Create dataset again
dates = pd.date_range(start='2024-01-01', periods=180, freq='D')

np.random.seed(42)

temperature = 20 + 10*np.sin(np.linspace(0, 3*np.pi, 180)) + np.random.normal(0, 2, 180)
humidity = 60 + 20*np.random.rand(180)
rainfall = np.random.choice([0, 0, 0, 5, 10, 20], size=180)

df = pd.DataFrame({
    'date': dates,
    'temperature': temperature,
    'humidity': humidity,
    'rainfall': rainfall
})

df.set_index('date', inplace=True)
```

---

## 3️⃣ Create Target Variable (VERY IMPORTANT)

```python
# Target = next day's temperature
df['target'] = df['temperature'].shift(-1)

df.head()
```

👉 Teach:

* We are predicting **future using past**
* `shift(-1)` → moves data upward

---

## 4️⃣ Create Features (Lag Features)

```python
# Past temperature
df['temp_lag_1'] = df['temperature'].shift(1)
df['temp_lag_2'] = df['temperature'].shift(2)
df['temp_lag_3'] = df['temperature'].shift(3)

# Past humidity
df['humidity_lag_1'] = df['humidity'].shift(1)

# Rolling feature
df['temp_rolling_3'] = df['temperature'].rolling(3).mean()

df.head()
```

---

## 5️⃣ Drop Missing Values

```python
df = df.dropna()
```

👉 Missing comes from:

* shifting
* rolling

---

## 6️⃣ Define Features (X) and Target (y)

```python
X = df[['temp_lag_1', 'temp_lag_2', 'temp_lag_3', 
        'humidity_lag_1', 'temp_rolling_3']]

y = df['target']
```

---

## 7️⃣ Train-Test Split (IMPORTANT RULE)

❌ Do NOT shuffle time series
✅ Split based on time

```python
split = int(len(df) * 0.8)

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]
```

---

## 8️⃣ Train Model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

---

## 9️⃣ Predictions

```python
y_pred = model.predict(X_test)
```

---

## 🔟 Evaluate Model

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

👉 Teach:

* Lower MAE = better predictions

---

## 1️⃣1️⃣ Compare Actual vs Predicted

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(y_test.values, label='Actual')
plt.plot(y_pred, label='Predicted')
plt.legend()
plt.title("Actual vs Predicted Temperature")
plt.show()
```

---

# 🎯 Key Concepts Students Learn

| Concept         | Why Important           |
| --------------- | ----------------------- |
| Lag features    | Capture past dependency |
| Rolling mean    | Smooth trend            |
| Target shifting | Predict future          |
| Time split      | Avoid data leakage      |
| Evaluation      | Measure accuracy        |

---

# 🧠 Core Insight (Very Important to Say)

> “We are not predicting randomly — we are using past patterns to estimate the future.”

---

# 💡 Upgrade Ideas (Next Level)

Once students understand this, you can extend:

### 🔹 Better Models

* Random Forest
* XGBoost

### 🔹 More Features

```python
df['month'] = df.index.month
df['day_of_week'] = df.index.dayofweek
```

### 🔹 Multi-step Forecast

* Predict next 7 days instead of 1

---

# 🧑‍🏫 Teaching Strategy

1. Ask:
   👉 “Can yesterday’s weather predict tomorrow?”

2. Build lag features LIVE

3. Train simple model

4. Show prediction graph

👉 This creates a **“wow moment”**

---

If you want next:
✅ Turn this into a **student assignment/project**
✅ Add **Random Forest / XGBoost version**
✅ Show **how to deploy this as an API (real-world use)**
````
