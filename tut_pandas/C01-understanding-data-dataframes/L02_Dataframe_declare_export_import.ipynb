```python
# Show on colab

import pandas as pd
import requests
from io import StringIO

url = 'https://en.wikipedia.org/wiki/List_of_countries_by_area'

# Add a User-Agent header to mimic a web browser and avoid 403 Forbidden errors
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}

# Fetch the HTML content using requests with the headers
response = requests.get(url, headers=headers)

# Pass the HTML content directly to pd.read_html wrapped in StringIO to address FutureWaring
tables = pd.read_html(StringIO(response.text))  # returns a list of DataFrames
print(f"Total tables: {len(tables)}")
df = tables[1] # Assuming the second table is usually the most relevant on such pages
print(df)
```
