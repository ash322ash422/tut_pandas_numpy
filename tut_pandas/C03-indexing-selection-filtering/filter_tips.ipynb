## On tips dataset
### Explanation of tips dataset

- **total_bill**: The total amount of the bill (numerical).
- **tip**: The amount of tip left by the customer (numerical).
- **sex**: The gender of the person paying the bill (categorical: **Male/Female**). 
- **smoker**: Indicates whether a smoker was present in the party (categorical: **Yes/No**).
- **day**: The day of the week when the meal occurred (categorical: **Thurs, Fri, Sat, Sun**).
- **time**: The time of day when the meal occurred (categorical: **Lunch/Dinner**).
- **size**: The number of people in the dining party (numerical).


```python
# Load the tips dataset

df = pd.read_csv("data_tips.csv")

print(df.head(10))
```

```python
# 1) Before filtering lets get to know all unique values of categorical columns
col = 'sex'
print(df[col].unique())
```

```python
# 2: 

cat_columns = ['sex', 'smoker', 'day', 'time', 'size']

for col in cat_columns:
    print(f"Column: {col}")
    print(df[col].unique())
    print("-----")
```

# Filter

## Filter: Customers who are females

```python
mask = (df['sex'] == 'Female')
# print(mask)
# print("----")

df_filter = df[mask]
print(df_filter.head()) # print first few records
print("#########################")
print("count:", len(df_filter.index)) # count of df_filter
```

```python
## Filter: Customers who are female smokers
```

```python

mask = (df['sex'] == 'Female') & (df['smoker'] == 'Yes')

df_filter = df[mask]
print(df_filter.head())
print("#########################")
print("count:", len(df_filter.index)) # count of df_filter
```

```python
## Filter: Customers whose total bill < 9 or > 50
```

```python

mask = (df['total_bill'] < 9) | (df['total_bill'] > 50)

df_filter = df[mask]
print(df_filter.head())
print("#########################")
print("count:", len(df_filter.index)) # count of df_filter
```

## Filter: Customers who tipped more than 20% of their total bill
- Here I do not have tip_percent column. So, I create one and then filter it out.

```python
df['tip_percent'] = (df['tip'] / df['total_bill']) * 100 # step1: create a column

mask = (df['tip_percent'] > 20) # step2: filter based on this new column

df_filter = df[mask]
print(df_filter.head())
print("#########################")
print("count:", len(df_filter.index)) # count of df_filter
```

## Filter: Lunch time male customers with total bill between 15 and 30

```python
mask = (df['time'] == 'Lunch') & (df['sex'] == 'Male') & (df['total_bill'].between(15, 30))

df_filter = df[mask]
print(df_filter.head())
print("#########################")
print("count:", len(df_filter.index)) # count of df_filter
```

## Filter: Female smokers who dined on weekends and gave tips over 4

```python
mask = (
    (df['sex'] == 'Female') &
    (df['smoker'] == 'Yes') &
    (df['day'].isin(['Sat', 'Sun'])) & # (df['day'] == 'Sat' |  df['day'] == 'Sun')
    (df['tip'] > 4)
)

df_filter = df[mask]
print(df_filter.head())
print("#########################")
print("count:", len(df_filter.index)) # count of df_filter
```

## Filter: Tables with size >= 4 but average tip per person < 1.5

```python
df['tip_per_person'] = df['tip'] / df['size'] # step1: create a new column

mask = (df['size'] >= 4) & (df['tip_per_person'] < 1.5) # step2: create mask

df_filter = df[mask]
print(df_filter.head())
print("#########################")
print("count:", len(df_filter.index)) # count of df_filter
```
